```python
# Program: Sort contents of a text file and write to another file

# Read input and output file names from the user
input_file = input("Enter the input text file name (with extension): ")
output_file = input("Enter the output text file name (with extension): ")

try:

    # Open the input file and read all lines
    with open(input_file, 'r') as file:
        lines = file.readlines()

    # Remove leading/trailing spaces and empty lines
    clean_lines = []
    for line in lines:
        line = line.strip()
        if len(line) > 0: # ignore empty lines
            clean_lines.append(line)

    # Sort the lines alphabetically
    clean_lines.sort()

    # Write the sorted lines to the output file
    with open(output_file, 'w') as file:
        for line in clean_lines:
            file.write(line + '\n')

    print(f"\nContents sorted successfully and written to '{output_file}'")

except FileNotFoundError:
    print("Error: The input file was not found.")
except Exception as e:
    print("An error occurred:", e)
```
