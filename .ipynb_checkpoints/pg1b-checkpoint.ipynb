```python
name=input("Enter your name:  ")
year_of_birth=int(input("Enter your year of birth:  "))
current_year=2025
age=current_year-year_of_birth
print(f"\nHello, {name}!")
print(f"You are {age} years old.")
if age>=60:
    print("You are a senior citizen.")
else:
    print("You aren't a senior citizen.")
```
