```python
# Program: Sort 6 subject marks using Bubble Sort (Descending order)

# Read 6 subject marks from the user
marks = []
print("Enter marks for 6 subjects:")
for i in range(6):
    mark = float(input(f"Subject {i+1} : "))
    marks.append(mark)

print("\nMarks entered:", marks)

# Bubble Sort algorithm (Descending order)
n = len(marks)
for i in range(n - 1):
    for j in range(0, n - i - 1):
        if marks[j] < marks[j + 1]: # Swap for descending order
            marks[j], marks[j + 1] = marks[j + 1], marks[j]

# Display the sorted marks
print("\n--- Student Marks Report ---")
print("Marks from highest to lowest:")
for mark in marks:
    print(mark)
```
