```python
#CREATE ANINITIAL LIST
my_list = []

print("Initial list:",my_list)

#insert elements into the list
element = input("\nEnter an element to insert :")
position = int(input("Enter the position to insert at (0-based index):"))
my_list.insert(position,element)
print("List after insertion:",my_list)

#APPEND AN ELEMENT TO THE LIST
element=input("\nEnter an element to append: ")
my_list.append(element)
print("List ahter apprnding:",my_list)

#Removing an Element from list
element = input("\nEnter an element to remove: ")
if element in my_list:
    my_list.remove(element)
    print("List after removing:",my_list)
else:
    print("Element not found in the list!")

#display lengthof the list
print("\nLength of the list:", len(my_list))

#Pop an element (Remove Last elemnt)
if my_list:
    popped=my_list.pop()
    print("Popped element:",popped)
    print("List after popping:",my_list)
else:
    print("List is empty,nothing to pop!")

#Clear the Entire list
my_list.clear()
print("\nList after clearing:",my_list)
```
